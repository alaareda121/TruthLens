```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
```

```python
train_path = '/kaggle/input/datasets/aryansingh16/deepfake-dataset/real_vs_fake/real-vs-fake/train'
valid_path = '/kaggle/input/datasets/aryansingh16/deepfake-dataset/real_vs_fake/real-vs-fake/valid'
test_path  = '/kaggle/input/datasets/aryansingh16/deepfake-dataset/real_vs_fake/real-vs-fake/test'
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# EfficientNet مش محتاج rescale
train_datagen = ImageDataGenerator(
    horizontal_flip=True,
    rotation_range=10,
    zoom_range=0.1,
    width_shift_range=0.1,
    height_shift_range=0.1
)
val_test_datagen = ImageDataGenerator()

train_gen = train_datagen.flow_from_directory(
    train_path, target_size=IMG_SIZE,
    batch_size=BATCH_SIZE, class_mode='binary'
)
valid_gen = val_test_datagen.flow_from_directory(
    valid_path, target_size=IMG_SIZE,
    batch_size=BATCH_SIZE, class_mode='binary'
)
test_gen = val_test_datagen.flow_from_directory(
    test_path, target_size=IMG_SIZE,
    batch_size=BATCH_SIZE, class_mode='binary',
    shuffle=False
)

print("Classes:", train_gen.class_indices)
print("Train:", train_gen.samples)
print("Valid:", valid_gen.samples)
print("Test :", test_gen.samples)
```

```python
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras.applications.efficientnet import preprocess_input

base_model = EfficientNetB3(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)
base_model.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3))
x = preprocess_input(inputs)        # ← ده المهم
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(1, activation='sigmoid')(x)

model = models.Model(inputs, outputs)
model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

early_stop = EarlyStopping(monitor='val_accuracy', patience=3, restore_best_weights=True)

history1 = model.fit(
    train_gen,
    validation_data=valid_gen,
    epochs=5,
    callbacks=[early_stop]
)
```

```python
base_model.trainable = True
for layer in base_model.layers[:-50]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

early_stop2 = EarlyStopping(
    monitor='val_accuracy', patience=4,
    restore_best_weights=True
)
checkpoint = ModelCheckpoint(
    'truthlens_best.h5',
    monitor='val_accuracy',
    save_best_only=True, verbose=1
)

print("=== Phase 2 Fine-tuning ===")
history2 = model.fit(
    train_gen,
    validation_data=valid_gen,
    epochs=15,
    callbacks=[early_stop2, checkpoint]
)
```

```python
test_gen.reset()
loss, acc = model.evaluate(test_gen)
print(f"\nTest Accuracy: {acc*100:.2f}%")

preds = (model.predict(test_gen) > 0.5).astype(int).flatten()
labels = test_gen.classes

print(classification_report(labels, preds, target_names=['fake', 'real']))

cm = confusion_matrix(labels, preds)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['fake','real'],
            yticklabels=['fake','real'])
plt.title('Confusion Matrix')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history1.history['accuracy'], label='Phase1 Train')
ax1.plot(history1.history['val_accuracy'], label='Phase1 Val')
ax1.plot(history2.history['accuracy'], label='Phase2 Train')
ax1.plot(history2.history['val_accuracy'], label='Phase2 Val')
ax1.set_title('Accuracy')
ax1.legend()

ax2.plot(history1.history['loss'], label='Phase1 Train')
ax2.plot(history1.history['val_loss'], label='Phase1 Val')
ax2.plot(history2.history['loss'], label='Phase2 Train')
ax2.plot(history2.history['val_loss'], label='Phase2 Val')
ax2.set_title('Loss')
ax2.legend()

plt.tight_layout()
plt.show()
```
